```python
# operasi elementwise (mengoperasikan setiap elemen yang berkorespondensi dengan elemen yang sama)
a = np.array([1, 2, 3, 4, 5])
b = np.array([6, 7, 8, 9, 10])

hasil1 = a + b
hasil2 = a - b
hasil3 = a * b
hasil4 = a / b
hasil5 = a ** 2

print("hasil 1 =", hasil1)
print("hasil 2 =", hasil2)
print("hasil 3 =", hasil3)
print("hasil 4 =", hasil4)
print("hasil 5 =", hasil5)
```

```python
# Misalkan gambarkan fungsi yang secara matematis ditulis sebagai berikut:
def f(x, y):
    return np.cos(np.exp(-x ** 2 - y ** 2)) * np.sin(np.sqrt(x ** 2 + y ** 2))

x = np.linspace(-np.pi, np.pi, 30)
y = np.linspace(-np.pi, np.pi, 30)

X, Y = np.meshgrid(x, y)
Z = f(X, Y)

fig = plt.figure()
ax = plt.axes(projection='3d')
ax.contour3D(X, Y, Z, 50, cmap='binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
```

```python
# Nomor 1: Matriks 5x5 Random (Max, Min, Mean)
print("--- NOMOR 1 ---")
# Membuat matriks 5x5 dengan nilai acak (misal 1 - 100)
matriks = np.random.randint(1, 100, size=(5, 5))

print("Matriks 5x5:")
print(matriks)

# Menghitung nilai maksimum, minimum, dan rata-rata
nilai_max = matriks.max()
nilai_min = matriks.min()
nilai_mean = matriks.mean()

print(f"\nNilai Maksimum : {nilai_max}")
print(f"Nilai Minimum  : {nilai_min}")
print(f"Nilai Rata-rata: {nilai_mean:.2f}\n")
```

```python
# Nomor 2: Grafik 2D Persamaan Implisit
# x^6 + 3x^4 y^2 - 3x^4 + 3x^2 y^4 - x^2 y^3 - 6x^2 y^2 + 3x^2 + y^6 - 3y^4 + 3y^2 = 1

x = np.linspace(-2, 2, 400)
y = np.linspace(-2, 2, 400)
X, Y = np.meshgrid(x, y)

# Definisi persamaan (Heart curve/Heart shape)
Z = (X**6 + 3*(X**4)*(Y**2) - 3*(X**4) + 3*(X**2)*(Y**4) 
     - (X**2)*(Y**3) - 6*(X**2)*(Y**2) + 3*(X**2) + Y**6 - 3*(Y**4) + 3*(Y**2))

plt.figure(figsize=(6, 6))
# Plot contour untuk Z = 1
plt.contour(X, Y, Z, levels=[1], colors='red')
plt.title('Grafik 2D Persamaan (Nomor 2)')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.show()
```

```python
# Nomor 3: Grafik 3D Persamaan z = (5y/4 - |x|)^2 + x^2

x3 = np.linspace(-3, 3, 100)
y3 = np.linspace(-3, 3, 100)
X3, Y3 = np.meshgrid(x3, y3)

# Definisi persamaan fungsi z
Z3 = (5*Y3/4 - np.abs(X3))**2 + X3**2

fig = plt.figure(figsize=(8, 6))
ax = plt.axes(projection='3d')

# Plot surface 3D
surf = ax.plot_surface(X3, Y3, Z3, cmap='viridis', edgecolor='none')
ax.set_title('Grafik 3D Persamaan (Nomor 3)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=5)
plt.show()
```
