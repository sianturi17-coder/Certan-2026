```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

- numpy dipakai untuk membuat array, pandas untuk membuat tabel (DataFrame).
- load_iris() memuat dataset Iris bawaan scikit-learn. Isinya 150 data bunga dengan 4 fitur (sepal length, sepal width, petal length, petal width) dan 3 kelas (setosa, versicolor, virginica).

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

- pd.DataFrame(iris.data, columns=iris.feature_names) mengubah data numerik menjadi tabel dengan nama kolom yang jelas.
- df["Species"] = iris.target_names[iris.target] menambah kolom Species berisi nama bunga (bukan angka 0, 1, 2).
- X adalah fitur (semua kolom kecuali Species), dan y adalah target/label yang mau diprediksi.
- print(X) menampilkan 150 baris x 4 kolom.

```python
y
```

Menampilkan label tiap data. Isinya masih teks (setosa, versicolor, virginica) dengan tipe object.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

- Algoritma machine learning bekerja dengan angka, jadi label teks diubah dengan LabelEncoder.
- fit_transform(y) mengubah setosa menjadi 0, versicolor menjadi 1, dan virginica menjadi 2.
- Objek lab disimpan supaya nanti hasil prediksi bisa dikembalikan ke nama species dengan inverse_transform.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

- train_test_split membagi data menjadi 80% data latih (120 data) dan 20% data uji (30 data).
- test_size=0.2 berarti 20% untuk data uji.
- random_state=0 membuat pembagian acak selalu sama, jadi hasilnya bisa direproduksi.
- Hasilnya adalah X_train, X_test, y_train, dan y_test.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

- Gaussian NB dipilih karena semua fitur Iris berupa angka kontinu yang diasumsikan berdistribusi normal.
- model.fit(X_train, y_train) melatih model. Model menghitung rata-rata (μ) dan simpangan baku (σ) tiap fitur untuk tiap kelas, serta probabilitas awal tiap kelas P(C).

```python
result = NB_model.predict(X_test)
result
```

- predict(X_test) memprediksi kelas untuk 30 data uji.
- Outputnya array angka 0, 1, 2 sesuai hasil encoding.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

- accuracy_score(y_test, result) membandingkan label asli (y_test) dengan hasil prediksi (result).
- Dikali 100 supaya jadi persen. Hasilnya 96,67%, artinya 29 dari 30 data diprediksi benar.

```python
#prediksi inputan data baru
#input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```
